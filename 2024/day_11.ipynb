```python
def behaviour_2(stone, past_results):
    if stone in past_results:
        return past_results[stone]

    if stone == 0:
        result = [1]
    else:
        n_digits = stone.bit_length() * 30103 // 100000 + 1  # ~ log10(stone)
        if n_digits % 2 == 0:
            div = 10 ** (n_digits // 2)
            result = [stone // div, stone % div]
        else:
            result = [stone * 2024]

    past_results[stone] = result
    return result
```

```python
def transform(stone, remaining_blink, memos):
    if remaining_blink == 0:
        return 1

    if stone in memos[remaining_blink]:
        return memos[remaining_blink][stone]

    transformed = []

    if stone == "0":
        transformed.append("1")
    elif len(stone) % 2 == 0:
        half = len(stone) // 2
        transformed.append(stone[:half])
        transformed.append(str(int(stone[half:])))
    else:
        transformed.append(str(int(stone) * 2024))

    res = sum(transform(new_stone, remaining_blink - 1,memos) for new_stone in transformed)

    memos[remaining_blink][stone] = res

    return res


def memoization(stones, blink):
    stones = [str(stone) for stone in stones]
    memos = [dict() for _ in range(blink + 1)]


    count = 0
    for stone in stones:
        count += transform(stone, blink, memos)

    return count
```
