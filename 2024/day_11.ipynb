```python
data = "4329 385 0 1444386 600463 19 1 56615"
stones = [int(x) for x in data.split()]
```

```python
stones
```

```python
def behaviour(x):
    if x == 0:
        return [1]
    n_digits = x.bit_length() * 30103 // 100000 + 1  # ~ log10(x), fast estimation
    if n_digits % 2 == 0:
        div = 10 ** (n_digits // 2)
        return [x // div, x % div]
    else:
        return [x * 2024]
```

### Answer 1

```python

import time

end_stone = stones.copy()

for i in range(25):
    start_time = time.time()
    end_stone = [x for stone in end_stone for x in behaviour(stone)]
    step_time = time.time() - start_time
    print(f"Step {i}: {step_time:.4f} seconds")

print("Final length:", len(end_stone))
```
