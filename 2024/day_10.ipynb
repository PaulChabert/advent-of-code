```python
def frame_grid(grid, frame=0):
    grid_c = [line.copy() for line in grid]
    grid_c.insert(0,[frame] * len(grid_c[0]))
    grid_c.append([frame] * len(grid_c[0]))
    [line.append(frame) for line in grid_c]
    [line.insert(0,frame) for line in grid_c]
    return grid_c
```

```python
def get_first_trails(f_grid):
    len_x = len(f_grid)
    len_y = len(f_grid[0])
    start_trails = []
    id = 0
    
    for x in range(1,len_x - 1):
        for y in range(1,len_y - 1):
            if f_grid[x][y] == 0:
                start_trails.append((0,x,y,id))
                id += 1

    return start_trails
```

```python
def success_neighbors(f_grid,trail):
    neighbors_coors = [(0,-1),(1,0),(0,1),(-1,0)]
    x,y = trail[1:3]
    success_set = set()
    for dx, dy in neighbors_coors:
        nx, ny = x + dx, y + dy
        score = trail[0]
        if f_grid[nx][ny] == score + 1:
            n_trail = (score + 1, nx, ny, trail[3])
            if n_trail not in success_set:
                success_set.add(n_trail)
    return success_set
```

```python
f_grid = frame_grid(grid)
hiking = set(get_first_trails(f_grid))
to_explore = list(hiking)

while to_explore:
    trail = to_explore.pop()
    for neighbor in success_neighbors(f_grid, trail):
        if neighbor not in hiking:
            hiking.add(neighbor)
            to_explore.append(neighbor)

len([trail for trail in hiking if trail[0] == 9])
```
