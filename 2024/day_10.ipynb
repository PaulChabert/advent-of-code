```python
data = """89010123
78121874
87430965
96549874
45678903
32019012
01329801
10456732"""
```

```python
data = [[int(char) for char in line.strip()] for line in data.split()]
grid = [line.copy() for line in data]
```

```python
def frame_grid(grid, frame=0):
    grid_c = [line.copy() for line in grid]
    grid_c.insert(0,[frame] * len(grid_c[0]))
    grid_c.append([frame] * len(grid_c[0]))
    [line.append(frame) for line in grid_c]
    [line.insert(0,frame) for line in grid_c]
    return grid_c
```

```python
def get_first_trails(f_grid):
    len_x = len(f_grid)
    len_y = len(f_grid[0])
    start_trails = []
    id = 0
    
    for x in range(1,len_x - 1):
        for y in range(1,len_y - 1):
            if f_grid[x][y] == 0:
                start_trails.append((0,(x,y),id))
                id += 1

    return start_trails
```

```python
get_first_trails(f_grid)
```

```python
def success_neighbors(f_grid,trail):
    neighbors_coors = [(0,-1),(1,0),(0,1),(-1,0)]
    x,y = trail['coors']
    seen = set()
    success_list = []
    for dx, dy in neighbors_coors:
        nx, ny = x + dx, y + dy
        if f_grid[nx][ny] == trail['score'] + 1:
            key = (trail['score'] + 1, (nx, ny), trail['id'])
            if key not in seen:
                seen.add(key)
                success_list.append({
                    'score': trail['score'] + 1,
                    'coors': (nx, ny),
                    'id': trail['id']
                })

    return success_list
```

```python
f_grid = frame_grid(grid)
hiking = get_first_trails(f_grid)

for trail in hiking:
    hiking.extend(success_neighbors(f_grid,trail))
```
