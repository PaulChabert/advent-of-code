# Part 1

```python
unzip_data = []

n = 0
for i,x in enumerate(data):
    if i % 2 == 0:
        unzip_data.extend([str(n)] * int(x))
        n += 1
    else:
        unzip_data.extend(['.'] * int(x))

list_unzip = list(unzip_data)
point_indexes = []
num_indexes = []

for i,x in enumerate(list_unzip):
    if x == ".":
        point_indexes.append((i,x))
    else:
        num_indexes.append((i,x))

for p,n in zip(point_indexes,num_indexes[::-1]):
    if n[0] < p[0]:
        break
    list_unzip[p[0]] = n[1]
    list_unzip[n[0]] = '.'

sum_is = sum(int(x) * i for i, x in enumerate(list_unzip) if x != ".")
print("Total sum:",sum_is)
```

```python
from collections import defaultdict

unzip_data = []
num_infos = []
point_infos = []

id = 0
fin_i = 0

for i, space in enumerate(data):
    space = int(space)
    if space == 0:
        continue
    if i % 2 == 0:
        unzip_data.extend([str(id)] * space)
        num_infos.append((fin_i, space))
        id += 1
    else:
        unzip_data.extend(['.'] * space)
        point_infos.append((fin_i, space))
    fin_i += space

# Work on copies
unzip_data_c = unzip_data[:]
num_infos_c = sorted(num_infos, reverse=True)
point_infos_c = point_infos[:]

for num_info in num_infos_c:
    src_start, src_len = num_info
    # Filter matching points only once
    matching_points = [pt for pt in point_infos_c if pt[0] < src_start and pt[1] >= src_len]

    if matching_points:
        best_point = min(matching_points, key=lambda x: x[0])
        dst_start = best_point[0]

        # Move block
        unzip_data_c[dst_start:dst_start + src_len] = unzip_data_c[src_start:src_start + src_len]
        unzip_data_c[src_start:src_start + src_len] = ['.'] * src_len

        # Update free space
        point_infos_c.remove(best_point)
        if best_point[1] > src_len:
            updated_point = (dst_start + src_len, best_point[1] - src_len)
            point_infos_c.append(updated_point)

# Final computation
total = sum(int(x) * i for i, x in enumerate(unzip_data_c) if x != '.')
print("Total sum:", total)
```
