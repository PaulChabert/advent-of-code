```python
from collections import defaultdict
from itertools import combinations

grid = [row.copy() for row in data]

atenna_position_dict = defaultdict(list)
antinodes_set = set()
len_l = len(grid) - 1
len_c = len(grid[0]) - 1
# Identify the antnna positions
for x,line in enumerate(grid):
    for y,frqc in enumerate(line):
        if frqc != '.':
            atenna_position_dict[frqc].append((x,y))

for frqc,coors in atenna_position_dict.items():
    for a, b in combinations(coors,2):
        x = 2 * a[0] - b[0]
        y = 2 * a[1] - b[1]
        if 0 <= x <= len_l and  0 <= y <= len_c:
            antinodes_set.add((x,y))
        x = 2 * b[0] - a[0]
        y = 2 * b[1] - a[1]
        if 0 <= x <= len_l and  0 <= y <= len_c:
            antinodes_set.add((x,y))

len(antinodes_set)
```

### Answer 2

```python
from collections import defaultdict
from itertools import combinations

grid = [row.copy() for row in data]

atenna_position_dict = defaultdict(list)
antinodes_set = set()
len_l = len(grid) - 1
len_c = len(grid[0]) - 1
# Identify the antnna positions
for x,line in enumerate(grid):
    for y,frqc in enumerate(line):
        if frqc != '.':
            atenna_position_dict[frqc].append((x,y))

for frqc,coors in atenna_position_dict.items():
    for a, b in combinations(coors,2):
        i = 0
        while True:
            x = (i + 1) * a[0] - i * b[0]
            y = (i + 1) * a[1] - i * b[1]
            if not(0 <= x <= len_l and  0 <= y <= len_c):
                break
            antinodes_set.add((x,y))
            i += 1
        i = 0
        while True:
            x = (i + 1) * b[0] - i * a[0]
            y = (i + 1) * b[1] - i * a[1]
            if not(0 <= x <= len_l and  0 <= y <= len_c):
                break
            antinodes_set.add((x,y))
            i += 1

len(antinodes_set)
```
